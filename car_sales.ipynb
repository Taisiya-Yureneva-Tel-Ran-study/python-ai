```python
import pandas as pd
df = pd.read_csv('car_sales_cleaned1.csv')
df
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df = df.dropna()
df.shape
```

```python
valuesCol = df['Model'].unique()
valuesCol  #.argsort()
```

```python
dfProjected = df[['Annual Income','Price ($)']]
dfProjected
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
X = dfProjected[['Price ($)']]
Y = dfProjected[['Annual Income']]
model.fit(X.values, Y)
predictions = model.predict([[30000]])
predictions
```
