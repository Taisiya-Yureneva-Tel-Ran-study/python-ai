```python
import numpy as np

docs = np.array([[0.3, -1.2, 0.5, 0.9],
               [0.7, 2.2, 0.3, 0.1],
               [1.92, -1.2, -0.7, 1.9],
               [0.32, 1.7, -0.5, -1.9],
               [4.3, 3.8, -0.1, 0.88]])

query = np.array([0.4, 1.8, -2.3, 0.1])

query_len = np.linalg.norm(query)
query_len
```

```python
q_norm = query / query_len
q_norm
```

```python
qq = np.linalg.norm(q_norm)
qq
```

```python
docs_len = np.linalg.norm(docs, axis=1, keepdims=True)
docs_len
```

```python
docs_norm = docs / docs_len
docs_norm
```

```python
similarities = docs_norm @ q_norm
similarities
```

```python
best_index_doc = np.argmax(similarities)
best_index_doc
```
